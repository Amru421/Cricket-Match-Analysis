# Cricket Data Analysis\n\nThis notebook explores the sample cricket dataset. The included records are fictional demonstration data.

```python
from pathlib import Path\nimport pandas as pd\nimport matplotlib.pyplot as plt\nimport seaborn as sns\n\nROOT = Path('..').resolve()\nDATA = ROOT / 'dataset'\nmatches = pd.read_csv(DATA / 'matches.csv')\nbatting = pd.read_csv(DATA / 'batting.csv')\nbowling = pd.read_csv(DATA / 'bowling.csv')\n\nprint('Matches:', len(matches))\ndisplay(matches.head())
```

## Batting performance

```python
batting['strike_rate'] = (batting['runs'] / batting['balls_faced'] * 100).round(2)\nbatting_summary = batting.groupby('player').agg(total_runs=('runs','sum'), balls=('balls_faced','sum'), dismissals=('dismissed','sum'), fours=('fours','sum'), sixes=('sixes','sum')).reset_index()\nbatting_summary['batting_average'] = (batting_summary['total_runs'] / batting_summary['dismissals'].replace(0, pd.NA)).round(2)\nbatting_summary['strike_rate'] = (batting_summary['total_runs'] / batting_summary['balls'] * 100).round(2)\ndisplay(batting_summary.sort_values('total_runs', ascending=False))
```

## Bowling performance

```python
bowling_summary = bowling.groupby('player').agg(overs=('overs','sum'), runs_conceded=('runs_conceded','sum'), wickets=('wickets','sum')).reset_index()\nbowling_summary['economy_rate'] = (bowling_summary['runs_conceded'] / bowling_summary['overs']).round(2)\nbowling_summary['bowling_average'] = (bowling_summary['runs_conceded'] / bowling_summary['wickets'].replace(0, pd.NA)).round(2)\ndisplay(bowling_summary.sort_values('wickets', ascending=False))
```

## Team results and charts

```python
team_wins = matches['winner'].value_counts().rename_axis('team').reset_index(name='wins')\ndisplay(team_wins)\nsns.set_theme(style='whitegrid')\nsns.barplot(data=batting_summary, x='player', y='total_runs')\nplt.title('Total Runs by Player')\nplt.show()\nsns.barplot(data=bowling_summary, x='player', y='wickets')\nplt.title('Total Wickets by Bowler')\nplt.show()\nsns.barplot(data=team_wins, x='team', y='wins')\nplt.title('Matches Won by Team')\nplt.show()
```
